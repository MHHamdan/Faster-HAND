# FasterHAND demo

Reads a handwritten document image with FasterHAND and returns one sequence of characters and layout
tokens (page, page number, section, annotation, body) in reading order. Four READ 2016 models are
available: single page, compact single page, double page and triple page.

**How to use:** run the three cells below in order (*Runtime → Run all*). The last cell prints a
public `gradio.live` link; open it, upload an image or pick an example, and press *Transcribe*.
The link stays valid while this notebook runs.

A GPU is optional: *Runtime → Change runtime type → T4 GPU* makes decoding faster, especially for
triple pages.

Code: [github.com/MHHamdan/Faster-HAND](https://github.com/MHHamdan/Faster-HAND) ·
Weights: [Hugging Face collection](https://huggingface.co/collections/MHamdan/hand-decoding-6ac083dab73470ecf7f95551)

## Using a model without the interface

```python
from hand_release.inference import HANDRecognizer

model = HANDRecognizer.from_pretrained("weights/hand-read2016-page", kv_cache=True, speculative=True)
page = model.read("demo/examples/read2016_page_test_11.jpg", source_dpi=150)
print(page.text)   # transcription; page.raw keeps the layout tokens inline
```

Pass `source_dpi=300` for a raw READ 2016 scan. Each model reads the number of pages it was
trained on.
